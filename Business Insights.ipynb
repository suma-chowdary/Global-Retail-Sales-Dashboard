# Business Insights – Global Retail Sales Dashboard

### OVERVIEW


##### Total Sales: 12.64M
##### Total Profit: 1.47M
##### Profit Margin: 11.61%
##### Return Rate: 5.95%
##### Sales Target: 15M
##### Target Achievement: 84.28%

### KEY BUSINESS INSIGHTS

### 1.Regional Profitability
Central region has the highest profit, while East and Africa have lower profitability.

### 2.Discount Impact
Higher discounts can reduce profitability, so discounts should be controlled.

### 3.Segment Performance
Consumer segment is the most profitable segment.

### 4.Returns
Central region has the highest number of returned orders and needs attention.

### 5.Revenue Concentration
A small group of top products contributes significantly to overall sales.

### 6.Shipping Cost
Standard Class is the most cost-effective shipping mode based on average shipping cost.

### 7.Manager Performance
Regional manager performance is not balanced, with differences in sales and profit.

### RECOMMENDATIONS

##### Improve sales performance to close the 15M target gap.
##### Control excessive discounts.
##### Focus on high-performing products and the Consumer segment.
##### Investigate reasons for high returns in Central region.
##### Use cost-effective shipping modes where appropriate.
##### Support weaker regions/managers with targeted strategies.
